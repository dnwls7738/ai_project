```python
from diffusers import StableDiffusionPipeline
import torch
from pathlib import Path
from tqdm import tqdm
import random
import shutil
```

```python
huggingface_token = "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"
```

```python
pipe = StableDiffusionPipeline.from_pretrained(
    "Lykon/dreamshaper-8",
    torch_dtype=torch.float16,
    use_auth_token=huggingface_token,
    use_dafetensors=True
).to("cuda")

pipe.enable_xformers_memory_efficient_attention()
pipe.enable_vae_slicing()
pipe.enable_vae_tiling()
```

```python
output_dir = Path("mask_images")
image_size = (640, 640)
batch_size = 16
images_per_class = 500
num_inference_steps = 25
```

```python
genders = [
    "asian man", "asian woman","white man", "white woman",
    "black man", "black woman","latino man", "latino woman",
    "middle eastern man", "middle eastern woman"
]
```

```python
poses = [
    "front view",
    "side view",
    "three-quarter view",
    "looking up",
    "looking down",
    "head tilted left",
    "head tilted right"
]
```

```python
background = [
    "indoor", "outdoor", "office", "street", "cafe",
    "school", "subway","park", "shopping mall",
    "restaurant", "library"
]
```

```python
clothes = [
    "in casual clothes", "wearing a hoodie", "in a business suit",
    "in a school uniform", "waring sportswear", "wearing traditional clothes"
]
```

```python
mask_colors = ["white","beige","blue","black","gray","pink"]
```

```python
expressions = [
    "smiling",
    "netural expression",
    "frowning",
    "surprised expression",
    "serious expression",
    "happy expression"
]
```

```python
shot_types = [
    "close-up face",
    "portrait, head and shoulders",
    "upper body portrait",
    "full body portrait"
]
```

```python
class_prompt_templates = {
    "mask_on":(
        "a {shot} of a {genders} wearing a {mask_colors} face mask covering mouth and nose properly,"
        "{expressions}, {poses}, {background}, {clothes}, realistic, high-quality"
    ),

    "no_mask" : (
        "a {shot} of a {genders} without mask, {expressions},{poses},"
        "{background}, {clothes}, realistic, high-quality"
    )
}
```

```python
negative_prompts = {
    "mask_on":"blurry, deformed face, low quality, unnatural skin",
    "no_mask":"blurry, deformed face, low quality, unnatural skin"
}
```

```python
for class_name, template in class_prompt_templates.items():
    class_folder = output_dir / class_name
    class_folder.mkdir(parents=True, exist_ok=True)

    for i in tqdm(range(0, images_per_class, batch_size), desc=class_name):
        prompts, negs = [], [negative_prompts[class_name]] * batch_size

        for _ in range(batch_size):
            prompt = template.format(
                shot=random.choice(shot_types),
                genders=random.choice(genders),
                poses=random.choice(poses),
                background=random.choice(background),
                clothes=random.choice(clothes),
                mask_colors=random.choice(mask_colors) if class_name != "no_mask" else "",
                expressions=random.choice(expressions)
            )
            prompts.append(prompt)
            
        with torch. inference_mode(), torch.autocast("cuda"):
            images =pipe(prompts, negative_prompt=negs,
                         height=image_size[1],
                         width=image_size[0],
                         num_inference_steps=num_inference_steps).images
        for idx, img in enumerate(images):
            img.save(class_folder /f"{class_name}_{i+idx:05}.png")
```
