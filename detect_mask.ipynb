```python
# 웹캠 영상과 화면 창을 다루기 위해 OpenCV를 불러옵니다.
import cv2
# 학습된 YOLO 모델을 사용하기 위해 Ultralytics의 YOLO 클래스를 불러옵니다.
from ultralytics import YOLO
```

```python
# 학습이 완료된 마스크 탐지 모델의 경로를 지정합니다.
model_path = r"C:/ai_project01/results/mask_detection/weights/best.pt"
# 지정한 가중치 파일로 YOLO 모델을 불러옵니다.
model = YOLO(model_path)
```

```python
# 기본 카메라 장치(보통 노트북 내장 카메라)를 엽니다.
cap = cv2.VideoCapture(0)
# 프로그램 종료 방법을 사용자에게 안내합니다.
print("Press 'q' to exit the program.")

# 카메라 프레임을 계속 읽으며 실시간 탐지를 수행합니다.
while True:
    # 카메라가 정상적으로 열리지 않았다면 반복을 종료합니다.
    if not cap.isOpened():
        print("Error: Could not open video stream.")
        break

    # 카메라에서 한 장의 프레임을 읽습니다.
    ret, frame = cap.read()
    # 프레임을 읽지 못했다면 반복을 종료합니다.
    if not ret:
        print("Error: Could not read frame.")
        break

    # 현재 프레임에서 신뢰도 0.5 이상인 마스크를 탐지합니다.
    results = model.predict(source=frame, conf=0.5, imgsz=640, verbose=False)
    # 탐지 상자와 클래스 이름을 원본 프레임에 그립니다.
    annotated_frame = results[0].plot()
    # 탐지 결과를 별도의 창에 표시합니다.
    cv2.imshow("Mask Detection", annotated_frame)

    # q 키를 누르면 실시간 탐지를 종료합니다.
    if cv2.waitKey(1) == ord('q'):
        break

# 카메라 장치를 해제합니다.
cap.release()
# OpenCV가 만든 모든 화면 창을 닫습니다.
cv2.destroyAllWindows()
```
